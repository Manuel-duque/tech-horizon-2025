```python
import matplotlib.pyplot as plt

# Datos
productos = ["Laptop", "Mouse", "Teclado", "Monitor"]
ventas = [45, 78, 62, 35]

# Gráfico de barras verticales
plt.bar(productos, ventas, color='skyblue')

plt.title("Ventas por Producto")
plt.xlabel("Producto")
plt.ylabel("Unidades Vendidas")
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# Colores diferentes para cada barra
colores = ['red', 'blue', 'green', 'orange']
plt.bar(productos, ventas, color=colores)

# Ancho de las barras
plt.bar(productos, ventas, width=0.5)

# Agregar valores sobre las barras
plt.bar(productos, ventas)
for i, v in enumerate(ventas):
    plt.text(i, v + 2, str(v), ha='center')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Generar datos de ejemplo: edades de 100 personas
edades = np.random.normal(30, 10, 100)

# Crear histograma
plt.hist(edades, bins=10, color='lightgreen', edgecolor='black')

plt.title("Distribución de Edades")
plt.xlabel("Edad")
plt.ylabel("Frecuencia")
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos: horas de estudio vs calificación
horas = [1, 2, 3, 4, 5, 6, 7, 8]
calificacion = [50, 60, 65, 70, 75, 85, 90, 95]

# Crear scatter plot
plt.scatter(horas, calificacion, color='purple', s=100)

plt.title("Horas de Estudio vs Calificación")
plt.xlabel("Horas de Estudio")
plt.ylabel("Calificación")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos de ejemplo
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 16]

categorias = ['A', 'B', 'C', 'D', 'E']
valores = [5, 7, 3, 8, 6]

datos = np.random.randn(100)  # 100 valores aleatorios para el histograma

# Crear figura con 2 filas, 2 columnas
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Gráfico 1: arriba izquierda [0, 0]
axes[0, 0].plot(x, y)
axes[0, 0].set_title("Línea")

# Gráfico 2: arriba derecha [0, 1]
axes[0, 1].bar(categorias, valores)
axes[0, 1].set_title("Barras")

# Gráfico 3: abajo izquierda [1, 0]
axes[1, 0].scatter(x, y)
axes[1, 0].set_title("Scatter")

# Gráfico 4: abajo derecha [1, 1]
axes[1, 1].hist(datos, bins=10)
axes[1, 1].set_title("Histograma")

plt.tight_layout()  # Evita superposición
plt.show()
```
