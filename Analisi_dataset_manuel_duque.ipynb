```python
import pandas as pd

# Cargar tu dataset
datos =  pd.read_csv("artists-data.csv")
# Mostrar primeras filas
print(datos.head())
# Mostrar cantidad de filas y columnas
print(f"Filas: {datos.shape[0]}, Columnas: {datos.shape[1]}")
# Eliminar datos nulos
datos = datos.dropna()
```

```python
# ANALISIS LOS DATOS
datos.info()
```

```python
datos.describe()
```

```python
# 1. Filtrar artistas con más de 200 canciones
filtro1 = datos[datos["Songs"] > 200]
print("Artistas con más de 200 canciones:")
print(filtro1.head())

# 2. Filtrar artistas cuyo género contenga 'Pop'
filtro2 = datos[datos["Genres"].str.contains("Pop", case=False, na=False)]
print("\nArtistas del género Pop:")
print(filtro2.head())

# 3. Filtrar artistas con popularidad mayor a 3
filtro3 = datos[datos["Popularity"] > 3]
print("\nArtistas con popularidad mayor a 3:")
print(filtro3.head())
```

```python
# Nuevas columnas
datos["Artist_length"] = datos["Artist"].astype(str).apply(len)
datos["Num_genres"] = datos["Genres"].astype(str).apply(lambda x: len(x.split(",")))

print(datos.head())
```
